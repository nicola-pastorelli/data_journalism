```python
import pandas as pd
import joblib
```

```python
def predict_magnitude(input_features):
    """
    Prevede la magnitudo di un evento sismico dato un dizionario di feature in input.

    Carica il preprocessore e il modello addestrato da file esterni, applica 
    la trasformazione dei dati di input e restituisce la previsione della magnitudo.

    Args:
        input_features (dict): Dizionario contenente le feature dell'evento da predire.

    Returns:
        float: Valore previsto della magnitudo.
    """
    # Caricamento del preprocessore serializzato (encoder, scalatori, ecc.)
    preprocessor = joblib.load("../encoder/preprocessor.pkl")
    # Caricamento del modello XGBoost addestrato
    model = joblib.load("../model/xgb_magnitude_model.pkl") 
    
    # Conversione del dizionario di input in DataFrame con una sola riga
    input_df = pd.DataFrame([input_features])
    
    # Applicazione della trasformazione di preprocessing al DataFrame di input
    processed_data = preprocessor.transform(input_df)
    
    # Restituisce la predizione del modello sul dato preprocessato (primo elemento dell'array)
    return model.predict(processed_data)[0]
```

```python

input_data = {
    "province": "FI",
    "depth": 10.0,
    "depth_category": "medium",
    "latitude": 43.7718,
    "longitude": 11.2541,
    "lat_sin": 0.691,
    "lat_cos": 0.723,
    "lon_sin": 0.195,
    "lon_cos": 0.981,
    "km": 5.0,
    "km_squared": 25.0,
    "direction_angle_sin": 0.0,
    "direction_angle_cos": 1.0,
    "is_night": 0,
    "season": "summer",
    "month_sin": 0.0,
    "month_cos": 1.0,
    "hour_sin": 0.0,
    "hour_cos": 1.0,
    "mag_mean_30d": 2.5,
    "mag_trend_30d": 0.1,
    "activity_ratio": 0.8,
    "last_event_km": 10.0
}

prediction = predict_magnitude(input_data)
print(f"Magnitudo predetta: {prediction}")
```
