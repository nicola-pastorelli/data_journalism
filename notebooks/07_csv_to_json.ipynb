```python
import csv
import json
```

```python
csv_file = "../data/terremoti_ingv_1985_2025_extended.csv"
json_file = "../data/terremoti.json"
```

```python
campi_utili = [
    "latitude", "longitude", "magnitude", "power", "place", "province_name", "region", "time"
]
```

```python
with open(csv_file, newline='', encoding='utf-8') as f:
    reader = csv.DictReader(f)  # Apre il file CSV e crea un lettore che restituisce ogni riga come dizionario (colonne come chiavi)
    dati = []  # Lista vuota per accumulare i dati processati

    for row in reader:
        try:
            # Per ogni riga, si cerca di costruire un dizionario con conversioni appropriate di tipo
            dati.append({
                "lat": float(row["latitude"]),           # Converte latitudine in float
                "lon": float(row["longitude"]),          # Converte longitudine in float
                "magnitude": float(row["magnitude"]),    # Converte magnitudo in float
                "power": row["power"],                    # Potenza (stringa)
                "place": row["place"],                    # Nome del luogo (stringa)
                "province": row["province_name"],        # Nome della provincia (stringa)
                "region": row["region"],                  # Regione (stringa)
                "time": row["time"],                      # Timestamp (stringa)
                "year": int(row["year"])                   # Anno convertito in intero
            })
        except (ValueError, KeyError):
            # Se manca un campo o non è convertibile al tipo richiesto, si ignora quella riga
            continue
```

```python
with open(json_file, 'w', encoding='utf-8') as f_out:

    # Scrive la lista di dizionari 'dati' in un file JSON minificato
    # ensure_ascii=False mantiene correttamente i caratteri speciali
    # separators elimina spazi e indentazione per ridurre le dimensioni del file

    json.dump(dati, f_out, ensure_ascii=False, separators=(',', ':'))

print(f"Salvati {len(dati)} eventi in {json_file}")
```
